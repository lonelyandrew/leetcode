# #206. Reverse Linked List

## Description

Given the head of a singly linked list, reverse the list, and return the *reversed* list.

**Example 1:**

![Example 1](image/206_1.png)

- Input: `head = [1,2,3,4,5]`
- Output: `[5,4,3,2,1]`

**Example 2:**

![Example 2](image/206_2.png)

- Input: `head = [1,2]`
- Output: `[2,1]`

**Example 3:**

- Input: `head = []`
- Output: `[]`

### Constraints:
- The number of nodes in the list is the range $[0, 5000]$.
- $-5000 <= Node.val <= 5000$

```python
from typing import Optional, List


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    @classmethod
    def from_list(cls, vals: List) -> Optional['ListNode']:
        head: Optional[ListNode] = None
        current: Optional[ListNode] = None
        for val in vals:
            if head is None:
                head = ListNode(val)
                current = head
            else:
                current.next = ListNode(val)
                current = current.next
        return head

    def to_list(self) -> List:
        result_list: List = []
        current: Optional[ListNode] = self
        while current is not None:
            result_list.append(current.val)
            current = current.next
        return result_list


class Solution:

    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        current, prev = head, None
        while current:
            current.next, current, prev = prev, current.next, current
        return prev
```

```python
# Case 1
head: Optional[ListNode] = ListNode.from_list([1, 2, 3, 4, 5])
assert Solution().reverseList(head).to_list() == [5, 4, 3, 2, 1]

# Case 2
head = ListNode.from_list([1, 2])
assert Solution().reverseList(head).to_list() == [2, 1]

# Case 3
assert Solution().reverseList(None) is None

print("Passed.")
```
